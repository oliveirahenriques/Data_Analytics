# 1. Filtragem de voos

Cada item abaixo utiliza uma condição específica. O método `query()` é apropriado quando a expressão é curta e legível; `isin()` é especialmente útil para testar pertencimento a um conjunto de valores.

### 1.1 Voos com atraso de chegada de duas horas ou mais

```python
flights.query("arr_delay >= 120")
```

A condição mantém os voos cujo atraso de chegada é igual ou superior a 120 minutos.

### 1.2 Voos para Houston, `IAH` ou `HOU`

```python
flights[flights["dest"].isin(["IAH", "HOU"])]
```

`isin()` retorna `True` quando o destino pertence à lista informada.

### 1.3 Voos operados por United, American ou Delta

```python
flights[flights["carrier"].isin(["UA", "AA", "DL"])]
```

Os códigos utilizados na base são `UA`, `AA` e `DL`.

### 1.4 Voos que partiram em julho, agosto ou setembro

```python
flights[flights["month"].between(7, 9)]
```

`between(7, 9)` inclui os limites e seleciona os meses 7, 8 e 9.

### 1.5 Voos que chegaram mais de duas horas atrasados, mas não partiram atrasados

```python
flights.query("arr_delay > 120 and dep_delay <= 0")
```

A condição combina atraso de chegada superior a 120 minutos com partida no horário ou antecipada.

### 1.6 Voos que partiram pelo menos uma hora atrasados, mas recuperaram mais de 30 minutos

```python
(
    flights
    .assign(gain=lambda d: d["dep_delay"] - d["arr_delay"])
    .query("dep_delay >= 60 and gain > 30")
)
```

Definimos `gain = dep_delay - arr_delay`. Um valor positivo indica que parte do atraso foi recuperada; `gain > 30` significa recuperação superior a 30 minutos.

# 2. Voos com os maiores atrasos de partida

```python
(
    flights
    .nlargest(10, "dep_delay")
    [["year", "month", "day", "carrier", "flight", "origin", "dest", "dep_delay"]]
)
```

`nlargest()` é uma forma curta e expressiva de retornar as linhas com os maiores valores de uma coluna.

# 3. Voos mais rápidos

```python
fastest_flights = (
    flights
    .dropna(subset=["distance", "air_time"])
    .assign(speed_mph=lambda d: d["distance"] / d["air_time"] * 60)
    .nlargest(10, "speed_mph")
)

fastest_flights[
    ["carrier", "flight", "origin", "dest", "distance", "air_time", "speed_mph"]
]
```

A velocidade média é calculada em milhas por hora. A multiplicação por 60 converte milhas por minuto em milhas por hora. Como valores extremos podem resultar de tempos de voo muito pequenos ou registros anômalos, eles devem ser investigados antes de qualquer conclusão substantiva.

# 4. Voos que viajaram mais longe

```python
flights.loc[
    flights["distance"].eq(flights["distance"].max()),
    ["carrier", "flight", "origin", "dest", "distance"],
]
```

A comparação com o valor máximo preserva todos os voos empatados na maior distância, ao contrário de uma seleção arbitrária de apenas uma linha.

# 5. A ordem entre `query()` e `sort_values()` importa?

```python
resultado_1 = (
    flights
    .query("dep_delay > 60")
    .sort_values("arr_delay", ascending=False)
)

resultado_2 = (
    flights
    .sort_values("arr_delay", ascending=False)
    .query("dep_delay > 60")
)

resultado_1.equals(resultado_2)
```

Quando o filtro não depende da posição das linhas, os dois pipelines tendem a produzir as mesmas linhas na mesma ordem. Entretanto, filtrar primeiro costuma ser mais eficiente, pois reduz o número de linhas que precisam ser ordenadas.

A equivalência deixa de valer quando há uma seleção por quantidade ou posição entre as operações. Por exemplo, “filtrar entre os dez maiores atrasos gerais” não é igual a “encontrar os dez maiores atrasos entre os voos filtrados”.

```python
# Dez maiores atrasos gerais, seguidos do filtro
flights.nlargest(10, "arr_delay").query("dep_delay > 60")
```

```python
# Dez maiores atrasos somente entre os voos filtrados
flights.query("dep_delay > 60").nlargest(10, "arr_delay")
```

# 6. Comparação entre `air_time` e `arr_time - dep_time`

```python
flights.assign(
    naive_duration=lambda d: d["arr_time"] - d["dep_time"]
)[["dep_time", "arr_time", "air_time", "naive_duration"]].head()
```

A subtração direta não funciona corretamente porque `dep_time` e `arr_time` são números no formato HHMM, não quantidades de minutos. Por exemplo, `1030 - 945` produz 85, embora a diferença entre 09:45 e 10:30 seja 45 minutos.

Além disso, alguns voos atravessam a meia-noite, origem e destino podem usar fusos horários diferentes, e `air_time` mede apenas o tempo no ar, enquanto a diferença entre chegada e partida inclui outras etapas operacionais.

```python
def hhmm_to_minutes(time):
    """Converte um horário numérico HHMM em minutos após a meia-noite."""
    return (time // 100) * 60 + time % 100

comparison = (
    flights
    .dropna(subset=["dep_time", "arr_time", "air_time"])
    .assign(
        dep_minutes=lambda d: hhmm_to_minutes(d["dep_time"]),
        arr_minutes=lambda d: hhmm_to_minutes(d["arr_time"]),
    )
    .assign(
        clock_elapsed=lambda d: (d["arr_minutes"] - d["dep_minutes"]) % 1440
    )
)

comparison[["dep_time", "arr_time", "air_time", "clock_elapsed"]].head()
```

Mesmo depois da conversão, `clock_elapsed` não precisa ser igual a `air_time`. Uma comparação rigorosa exigiria datetimes completos, fusos horários dos aeroportos e conversão para uma referência temporal comum, como UTC.

# 7. Relação entre `dep_time`, `sched_dep_time` e `dep_delay`

Em princípio, `dep_delay` corresponde à diferença, em minutos, entre o horário real e o horário programado de partida. Valores positivos indicam atraso, zero indica partida no horário e valores negativos indicam partida antecipada.

```python
departure_check = (
    flights
    .dropna(subset=["dep_time", "sched_dep_time", "dep_delay"])
    .assign(
        actual=lambda d: hhmm_to_minutes(d["dep_time"]),
        scheduled=lambda d: hhmm_to_minutes(d["sched_dep_time"]),
    )
    .assign(
        calculated_delay=lambda d: (d["actual"] - d["scheduled"] + 720) % 1440 - 720
    )
)

departure_check[
    ["dep_time", "sched_dep_time", "dep_delay", "calculated_delay"]
].head()
```

```python
(
    departure_check
    .assign(difference=lambda d: d["dep_delay"] - d["calculated_delay"])
    ["difference"]
    .value_counts()
    .head()
)
```

A normalização modular trata melhor horários próximos da meia-noite e permite verificar se a diferença calculada coincide com `dep_delay`.

# 12. Qual companhia apresenta os piores atrasos?

A expressão “pior atraso” precisa ser definida. A média é sensível a extremos; a mediana representa melhor o voo típico; e a proporção de atrasos superiores a 15 minutos mede a frequência de atrasos relevantes.

```python
carrier_delays = (
    flights
    .groupby("carrier", as_index=False)
    .agg(
        flights=("flight", "size"),
        observed_delays=("arr_delay", "count"),
        mean_delay=("arr_delay", "mean"),
        median_delay=("arr_delay", "median"),
        late_share=("arr_delay", lambda x: x.gt(15).mean()),
    )
    .query("observed_delays >= 100")
    .sort_values("mean_delay", ascending=False)
)

carrier_delays
```

A classificação pode mudar conforme a métrica. Além disso, companhias operam combinações diferentes de origens, destinos, horários e distâncias. Portanto, uma comparação bruta não identifica isoladamente um efeito da companhia.

```python
carrier_dest = (
    flights
    .groupby(["carrier", "dest"], as_index=False)
    .agg(
        flights=("flight", "size"),
        mean_delay=("arr_delay", "mean"),
    )
    .query("flights >= 30")
)

carrier_dest.head()
```

```python
carrier_adjusted = (
    carrier_dest
    .assign(weighted_delay=lambda d: d["mean_delay"] * d["flights"])
    .groupby("carrier", as_index=False)
    .agg(
        weighted_delay=("weighted_delay", "sum"),
        flights=("flights", "sum"),
    )
    .assign(adjusted_mean=lambda d: d["weighted_delay"] / d["flights"])
    .sort_values("adjusted_mean", ascending=False)
)

carrier_adjusted
```

O agrupamento por companhia e destino reduz parte da diferença de composição, mas não controla completamente origem, horário, mês, distância, congestionamento ou condições meteorológicas. Uma separação mais rigorosa exigiria um modelo estatístico com esses fatores como controles.

# 13. Voo mais atrasado para cada destino

```python
most_delayed_by_dest = (
    flights
    .dropna(subset=["arr_delay"])
    .loc[lambda d: d.groupby("dest")["arr_delay"].idxmax()]
    .sort_values("arr_delay", ascending=False)
)

most_delayed_by_dest[
    ["dest", "carrier", "flight", "origin", "arr_delay", "year", "month", "day"]
]
```

`idxmax()` encontra o índice do maior atraso de cada destino. Em caso de empate, retorna apenas a primeira ocorrência.

### Preservando todos os empates

```python
most_delayed_with_ties = flights.loc[
    flights["arr_delay"].eq(
        flights.groupby("dest")["arr_delay"].transform("max")
    )
].sort_values(["dest", "arr_delay"], ascending=[True, False])

most_delayed_with_ties
```

# 14. Variação dos atrasos ao longo do dia

```python
hourly_delays = (
    flights
    .groupby("hour", as_index=False)
    .agg(
        flights=("flight", "size"),
        observed_delays=("dep_delay", "count"),
        mean_dep_delay=("dep_delay", "mean"),
        median_dep_delay=("dep_delay", "median"),
        mean_arr_delay=("arr_delay", "mean"),
        late_share=("dep_delay", lambda x: x.gt(15).mean()),
    )
)

hourly_delays
```

A análise considera média, mediana, proporção de voos com atraso superior a 15 minutos e quantidade de observações. A média pode ser influenciada por poucos atrasos extremos; a mediana e a proporção oferecem visões complementares.

```python
sns.lineplot(
    data=hourly_delays,
    x="hour",
    y="mean_dep_delay",
    marker="o",
)

plt.xlabel("Hora programada de partida")
plt.ylabel("Atraso médio na partida, em minutos")
plt.title("Atraso médio ao longo do dia")
plt.show()
```

### Comparação entre média e mediana

```python
hourly_long = hourly_delays.melt(
    id_vars="hour",
    value_vars=["mean_dep_delay", "median_dep_delay"],
    var_name="statistic",
    value_name="delay",
)

sns.lineplot(
    data=hourly_long,
    x="hour",
    y="delay",
    hue="statistic",
    marker="o",
)

plt.xlabel("Hora programada de partida")
plt.ylabel("Atraso na partida, em minutos")
plt.title("Média e mediana do atraso ao longo do dia")
plt.show()
```

O resultado descreve uma associação entre hora programada e atraso. Ele não demonstra, isoladamente, que a hora do dia cause o atraso, pois outras características dos voos também variam ao longo do dia.
